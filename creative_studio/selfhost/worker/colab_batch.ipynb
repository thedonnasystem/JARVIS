# FENDEZY creative batch (Google Colab)

Colab's terms do not allow a notebook to serve a web service, connect to remote proxies or run a remote worker
(research.google.com/colaboratory/faq.html, checked 2026-11-23). So this notebook does NOT start a tunnel: it runs a batch
file inside the notebook and gives you a zip of the outputs with their metadata (`ai_generated: true`).

1. **Runtime > Change runtime type > T4 GPU**.
2. **Run All**, then upload a batch file when asked (a JSON list; example in the cell below).
3. Download `fendezy_batch.zip` when it finishes. Import the files into the studio with `creative_owner_import` (owner step).

```python
BRANCH = "build/creative-selfhost"   # change to "main" after the merge
EXAMPLE = [{"module": "image", "model": "z-image-turbo", "prompt": "a ceramic coffee mug on a wooden table, soft daylight", "aspect": "1:1"},
           {"module": "voice", "model": "kokoro-82m", "script": "New drop this Friday."}]
import json, os
from google.colab import files
up = files.upload()
name = next(iter(up)) if up else None
if not name:
    name = "batch.json"
    json.dump(EXAMPLE, open(name, "w"))
os.environ["FZ_BATCH"] = os.path.abspath(name)
```

```python
!rm -rf /content/fz/repo && git clone -q --depth 1 -b $BRANCH https://github.com/thedonnasystem/JARVIS /content/fz/repo
!python3 /content/fz/repo/creative_studio/selfhost/worker/fz_worker.py --profile t4 --tunnel none --batch "$FZ_BATCH" --root /content/fz --out /content/fz/out
!cd /content/fz && zip -qr fendezy_batch.zip out
from google.colab import files
files.download("/content/fz/fendezy_batch.zip")
```
